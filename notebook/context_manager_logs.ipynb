```python
# Import the necessary libraries
import psycopg2
from IPython.display import display, Markdown
import matplotlib.pyplot as plt

# Define the database connection parameters
db_params = {
    'host': '127.0.0.1',
    'port': '5432',
    'database': 'solorpg',
    'user': 'solorpg',
    'password': 'solorpg'
}

# Connect to the database
conn = psycopg2.connect(**db_params)

# Create a cursor object to interact with the database
cursor = conn.cursor()

# Execute a query to fetch data from the context_manager_logs table
query = "SELECT * FROM context_manager_logs ORDER BY created_at DESC LIMIT 30"
cursor.execute(query)

# Fetch all the rows returned by the query
rows = cursor.fetchall()
rows.reverse()

# Assuming you have fetched the rows into a variable called 'rows'

# Extract the relevant columns from the rows
system_actual_size = [row[2] for row in rows]
system_clipped_size = [row[3] for row in rows]
lore_actual_size = [row[4] for row in rows]
lore_clipped_size = [row[5] for row in rows]
active_scene_actual_size = [row[6] for row in rows]
active_scene_clipped_size = [row[7] for row in rows]
recent_turns_actual_size = [row[8] for row in rows]
recent_turns_clipped_size = [row[9] for row in rows]
misc_actual_size = [row[10] for row in rows]
misc_clipped_size = [row[11] for row in rows]


# Calculate the total length of each row
total_length = sum(system_actual_size) + sum(lore_actual_size) + sum(active_scene_actual_size) + sum(recent_turns_actual_size) + sum(misc_actual_size)

# Calculate the percentage of each part
system_percentage = (sum(system_actual_size) / total_length) * 100
lore_percentage = (sum(lore_actual_size) / total_length) * 100
active_scene_percentage = (sum(active_scene_actual_size) / total_length) * 100
recent_turns_percentage = (sum(recent_turns_actual_size) / total_length) * 100
misc_percentage = (sum(misc_actual_size) / total_length) * 100

# Create a pie chart
labels = ['System', 'Lore', 'Active Scene', 'Recent Turns', 'Misc']
sizes = [system_percentage, lore_percentage, active_scene_percentage, recent_turns_percentage, misc_percentage]
colors = ['red', 'green', 'blue', 'orange', 'purple']
explode = (0, 0, 0, 0, 0)  # Adjust the explode values to create space between the slices

fig1, ax1 = plt.subplots()
ax1.pie(sizes, labels=labels, colors=colors, explode=explode, autopct='%1.1f%%', startangle=90)
ax1.axis('equal')

# Set the title
ax1.set_title('Percentage of actual Sizes')

# Show the chart
plt.show()

# create a line chart
plt.figure(figsize=(10, 6))
plt.plot(system_actual_size, label='System')
plt.plot(lore_actual_size, label='Lore')
plt.plot(active_scene_actual_size, label='Active Scene')
plt.plot(recent_turns_actual_size, label='Recent Turns')
plt.plot(misc_actual_size, label='Misc')

# add labels and title
plt.xlabel('Time')
plt.ylabel('actual Size')
plt.title('Line Chart of actual Sizes Over Time')

# add legend
plt.legend()

# show the chart
plt.show()

# Show the chart
plt.figure(figsize=(10, 6))
plt.plot(system_clipped_size, label='System')
plt.plot(lore_clipped_size, label='Lore')
plt.plot(active_scene_clipped_size, label='Active Scene')
plt.plot(recent_turns_clipped_size, label='Recent Turns')
plt.plot(misc_clipped_size, label='Misc')

# add labels and title
plt.xlabel('Time')
plt.ylabel('clipped Size')
plt.title('Line Chart of Clipped Sizes Over Time')

# add legend
plt.legend()

# show the chart
plt.show()

# Close the cursor and the database connection
cursor.close()
conn.close()
```
